```python
import cv2
import numpy as np
from collections import Counter


def generate_scene_img(path="/content/test.jpeg"):
    original_image_path = path
    original_image = cv2.imread(original_image_path)
    img_height, img_width, _ = original_image.shape
    new_image = np.ones((img_height, img_width, 3), dtype=np.uint8) * 255
    pixels = original_image.reshape(-1, 3)
    unique_colors, counts = np.unique(pixels, axis=0, return_counts=True)
    bg_color_arr = unique_colors[np.argmax(counts)]
    bg_color = tuple(int(x) for x in bg_color_arr)
    assigned_mask = np.zeros((img_height, img_width), dtype=bool)
    names = ["circle", "square", "triangle", "rectangle", "ellipse", "pentagon", "polygon"]

    def get_dominant_color(region, mask):
        pixels_valid = region[mask]
        if len(pixels_valid) == 0:
            return (0, 0, 0)
        colors = [tuple(pixel) for pixel in pixels_valid]
        dominant = Counter(colors).most_common(1)[0][0]
        return dominant

    boxes = list(result[0].boxes)
    boxes_sorted = sorted(boxes, key=lambda b: b.conf)

    for box in boxes_sorted:
        shape_index = int(box.cls[0])
        shape_name = names[shape_index]
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        bbox_w = x2 - x1
        bbox_h = y2 - y1
        shape_mask = np.zeros((bbox_h, bbox_w), dtype=np.uint8)
        if shape_name == "circle":
            center = (bbox_w // 2, bbox_h // 2)
            radius = min(bbox_w, bbox_h) // 2
            cv2.circle(shape_mask, center, radius, 255, -1)
        elif shape_name in ["square", "rectangle"]:
            cv2.rectangle(shape_mask, (0, 0), (bbox_w, bbox_h), 255, -1)
        elif shape_name == "triangle":
            pts = np.array([[0, bbox_h], [bbox_w // 2, 0], [bbox_w, bbox_h]], np.int32)
            cv2.fillPoly(shape_mask, [pts], 255)
        elif shape_name == "ellipse":
            center = (bbox_w // 2, bbox_h // 2)
            axes = (bbox_w // 2, bbox_h // 2)
            cv2.ellipse(shape_mask, center, axes, 0, 0, 360, 255, -1)
        elif shape_name == "pentagon":
            center = (bbox_w / 2, bbox_h / 2)
            radius = min(bbox_w, bbox_h) / 2
            pts = []
            start_angle = -np.pi/2
            for i in range(5):
                angle = start_angle + i * (2 * np.pi / 5)
                x = int(center[0] + radius * np.cos(angle))
                y = int(center[1] + radius * np.sin(angle))
                pts.append([x, y])
            pts = np.array(pts, np.int32)
            cv2.fillPoly(shape_mask, [pts], 255)
        elif shape_name == "polygon":
            center = (bbox_w / 2, bbox_h / 2)
            radius = min(bbox_w, bbox_h) / 2
            pts = []
            start_angle = -np.pi/2
            for i in range(8):
                angle = start_angle + i * (2 * np.pi / 8)
                x = int(center[0] + radius * np.cos(angle))
                y = int(center[1] + radius * np.sin(angle))
                pts.append([x, y])
            pts = np.array(pts, np.int32)
            cv2.fillPoly(shape_mask, [pts], 255)
        region = original_image[y1:y2, x1:x2]
        assigned_region = assigned_mask[y1:y2, x1:x2]
        shape_bool = (shape_mask == 255)
        bg_color_array = np.array(bg_color, dtype=np.uint8)
        not_bg = ~np.all(region == bg_color_array, axis=-1)
        valid_mask = shape_bool & (~assigned_region) & not_bg
        dominant_color = get_dominant_color(region, valid_mask)
        colored_shape = np.zeros((bbox_h, bbox_w, 3), dtype=np.uint8)
        colored_shape[:] = dominant_color
        new_region = new_image[y1:y2, x1:x2]
        new_region[shape_bool] = colored_shape[shape_bool]
        new_image[y1:y2, x1:x2] = new_region
        assigned_region[shape_bool] = True
        assigned_mask[y1:y2, x1:x2] = assigned_region

    output_path = "/content/generated_shapes.jpeg"
    cv2.imwrite(output_path, new_image)
    print(f"Imagen generada guardada en {output_path}")


generate_scene_img(
    path="/content/test.jpeg"
)
```

```python
import cv2
import numpy as np
import json
from collections import Counter
import math

def generate_scene_json(path="/content/test.jpeg"):
    result = model.predict(path)
    original_image_path = path
    original_image = cv2.imread(original_image_path)
    img_height, img_width, _ = original_image.shape
    pixels = original_image.reshape(-1, 3)
    unique_colors, counts = np.unique(pixels, axis=0, return_counts=True)
    bg_color_arr = unique_colors[np.argmax(counts)]
    bg_color = tuple(int(x) for x in bg_color_arr)
    assigned_mask = np.zeros((img_height, img_width), dtype=bool)
    names = ["circle", "square", "triangle", "rectangle", "ellipse", "pentagon", "polygon"]

    def get_dominant_color(region, mask):
        pixels_valid = region[mask]
        if len(pixels_valid) == 0:
            return (0, 0, 0)
        colors = [tuple(pixel) for pixel in pixels_valid]
        dominant = Counter(colors).most_common(1)[0][0]
        return dominant

    def get_geometry(shape_name, x1, y1, x2, y2):
        bbox_w = x2 - x1
        bbox_h = y2 - y1
        if shape_name == "circle":
            center = {"x": (x1 + x2) // 2, "y": (y1 + y2) // 2}
            radius = min(bbox_w, bbox_h) // 2
            return {"center": center, "radius": radius}
        elif shape_name == "ellipse":
            center = {"x": (x1 + x2) // 2, "y": (y1 + y2) // 2}
            radii = {"width": bbox_w // 2, "height": bbox_h // 2}
            return {"center": center, "radii": radii}
        elif shape_name == "triangle":
            pts = [{"x": x1, "y": y2}, {"x": (x1 + x2) // 2, "y": y1}, {"x": x2, "y": y2}]
            return pts
        elif shape_name in ["square", "rectangle"]:
            if bbox_w == bbox_h:
                return {"topleft": {"x": x1, "y": y1}, "side": bbox_w}
            else:
                sides = {"width": bbox_w, "height": bbox_h}
                return {"topleft": {"x": x1, "y": y1}, "sides": sides}
        elif shape_name in ["polygon", "pentagon"]:
            num_vertices = 5 if shape_name == "pentagon" else 8
            center_x = x1 + bbox_w / 2
            center_y = y1 + bbox_h / 2
            radius = min(bbox_w, bbox_h) / 2
            pts = []
            start_angle = -math.pi / 2
            for i in range(num_vertices):
                angle = start_angle + i * (2 * math.pi / num_vertices)
                px = int(center_x + radius * math.cos(angle))
                py = int(center_y + radius * math.sin(angle))
                pts.append({"x": px, "y": py})
            return pts

    boxes = list(result[0].boxes)
    boxes_sorted = sorted(boxes, key=lambda b: b.conf)
    objects = []
    for i, box in enumerate(boxes_sorted):
        shape_index = int(box.cls[0])
        shape_name = names[shape_index]
        json_type = "polygon" if shape_name == "pentagon" else shape_name
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        bbox_w = x2 - x1
        bbox_h = y2 - y1
        shape_mask = np.zeros((bbox_h, bbox_w), dtype=np.uint8)
        if shape_name == "circle":
            center = (bbox_w // 2, bbox_h // 2)
            radius = min(bbox_w, bbox_h) // 2
            cv2.circle(shape_mask, center, radius, 255, -1)
        elif shape_name in ["square", "rectangle"]:
            cv2.rectangle(shape_mask, (0, 0), (bbox_w, bbox_h), 255, -1)
        elif shape_name == "triangle":
            pts = np.array([[0, bbox_h], [bbox_w // 2, 0], [bbox_w, bbox_h]], np.int32)
            cv2.fillPoly(shape_mask, [pts], 255)
        elif shape_name == "ellipse":
            center = (bbox_w // 2, bbox_h // 2)
            axes = (bbox_w // 2, bbox_h // 2)
            cv2.ellipse(shape_mask, center, axes, 0, 0, 360, 255, -1)
        elif shape_name == "pentagon":
            center_rel = (bbox_w / 2, bbox_h / 2)
            radius_val = min(bbox_w, bbox_h) / 2
            pts = []
            start_angle = -np.pi / 2
            for j in range(5):
                angle = start_angle + j * (2 * np.pi / 5)
                x_rel = int(center_rel[0] + radius_val * np.cos(angle))
                y_rel = int(center_rel[1] + radius_val * np.sin(angle))
                pts.append([x_rel, y_rel])
            pts = np.array(pts, np.int32)
            cv2.fillPoly(shape_mask, [pts], 255)
        elif shape_name == "polygon":
            center_rel = (bbox_w / 2, bbox_h / 2)
            radius_val = min(bbox_w, bbox_h) / 2
            pts = []
            start_angle = -np.pi / 2
            for j in range(8):
                angle = start_angle + j * (2 * np.pi / 8)
                x_rel = int(center_rel[0] + radius_val * np.cos(angle))
                y_rel = int(center_rel[1] + radius_val * np.sin(angle))
                pts.append([x_rel, y_rel])
            pts = np.array(pts, np.int32)
            cv2.fillPoly(shape_mask, [pts], 255)
        region = original_image[y1:y2, x1:x2]
        assigned_region = assigned_mask[y1:y2, x1:x2]
        shape_bool = (shape_mask == 255)
        bg_color_array = np.array(bg_color, dtype=np.uint8)
        not_bg = ~np.all(region == bg_color_array, axis=-1)
        valid_mask = shape_bool & (~assigned_region) & not_bg
        dominant_color = get_dominant_color(region, valid_mask)
        dom_color_rgb = {"red": int(dominant_color[2]), "green": int(dominant_color[1]), "blue": int(dominant_color[0])}
        geometry = get_geometry(shape_name, x1, y1, x2, y2)
        obj = {
            "id": i + 1,
            "type": json_type,
            "color": dom_color_rgb,
            "points": geometry
        }
        objects.append(obj)
        assigned_region[shape_bool] = True
        assigned_mask[y1:y2, x1:x2] = assigned_region

    relations = []
    for j in range(1, len(objects)):
        relations.append({"obj1": objects[j - 1]["id"], "obj2": objects[j]["id"]})

    scene = {
        "size": {"width": img_width, "height": img_height},
        "background": {"red": bg_color[2], "green": bg_color[1], "blue": bg_color[0]},
        "objects": objects,
        "relations": relations
    }
    return json.dumps(scene, indent=2)

print(generate_scene_json(
    path="/content/test.jpeg"
))
```

```python
import cv2
import numpy as np
import json

def draw_scene_from_json(scene_json):
    if isinstance(scene_json, str):
        scene = json.loads(scene_json)
    else:
        scene = scene_json

    width = scene["size"]["width"]
    height = scene["size"]["height"]
    bg = scene["background"]
    bg_color = (bg["blue"], bg["green"], bg["red"])
    image = np.full((height, width, 3), bg_color, dtype=np.uint8)

    objects = sorted(scene["objects"], key=lambda obj: obj["id"])
    for obj in objects:
        typ = obj["type"]
        col = obj["color"]
        color = (col["blue"], col["green"], col["red"])
        pts = obj["points"]
        if typ == "circle":
            center = pts["center"]
            radius = pts["radius"]
            cv2.circle(image, (center["x"], center["y"]), radius, color, thickness=-1)
        elif typ == "ellipse":
            center = pts["center"]
            radii = pts["radii"]
            cv2.ellipse(image, (center["x"], center["y"]), (radii["width"], radii["height"]), 0, 0, 360, color, thickness=-1)
        elif typ in ["triangle", "polygon"]:
            points = np.array([[p["x"], p["y"]] for p in pts], dtype=np.int32)
            cv2.fillPoly(image, [points], color)
        elif typ == "rectangle":
            topleft = pts["topleft"]
            sides = pts["sides"]
            x, y = topleft["x"], topleft["y"]
            w, h = sides["width"], sides["height"]
            cv2.rectangle(image, (x, y), (x+w, y+h), color, thickness=-1)
        elif typ == "square":
            topleft = pts["topleft"]
            side = pts["side"]
            x, y = topleft["x"], topleft["y"]
            cv2.rectangle(image, (x, y), (x+side, y+side), color, thickness=-1)
    return image


scene_json_str = generate_scene_json()
drawn_image = draw_scene_from_json(scene_json_str)
cv2.imwrite("/content/drawn_scene.jpeg", drawn_image)
```
